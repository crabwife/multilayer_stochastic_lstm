```python
import json
from pathlib import Path
import numpy as np
import pandas as pd
```

```python
SRC=Path.cwd().absolute(); TASK=SRC.parent; ROOT=TASK.parent
INPUT=TASK/'input'; OUTPUT=TASK/'output'
CFG=json.loads((ROOT/'config.json').read_text()); SEED=CFG['seed']; np.random.seed(SEED)
```

```python
from import_data import fetch, save, DATASET
```

```python
import hashlib
cached=(OUTPUT/"manifest.json").exists() and (OUTPUT/"e001_raw.csv").exists() and not list(INPUT.glob("*.csv"))
if cached:
    raw=(OUTPUT/"e001_raw.csv").read_bytes()
    provenance=json.loads((OUTPUT/"manifest.json").read_text())
    cached=(provenance.get("dataset")==DATASET and provenance.get("sha256")==hashlib.sha256(raw).hexdigest())
if not cached:
    raw, provenance=fetch(INPUT)
```

```python
provenance=save(raw, provenance, OUTPUT)
(OUTPUT/'manifest.json').write_text(json.dumps(provenance, indent=2))
display(pd.DataFrame([provenance]))
```
